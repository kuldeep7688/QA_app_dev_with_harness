```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

NVIDIA_API_KEY = os.getenv("NVIDIA_API_KEY")
print(f"API Key loaded: {'Yes' if NVIDIA_API_KEY else 'No'}")

# Initialize OpenAI client with NVIDIA's base URL
client = OpenAI(
    base_url="https://integrate.api.nvidia.com/v1",
    api_key=NVIDIA_API_KEY
)
```

```python
def generate_poem(prompt: str, max_tokens: int = 200, temperature: float = 0.8) -> str:
    """
    Generate a poem using Gemma-2-2B-IT via NVIDIA's API (OpenAI-compatible).
    
    Args:
        prompt: The prompt to generate a poem from
        max_tokens: Maximum tokens to generate
        temperature: Sampling temperature (0.0-1.0)
    
    Returns:
        Generated poem text
    """
    completion = client.chat.completions.create(
        model="google/gemma-2-2b-it",
        messages=[
            {
                "role": "assistant",
                "content": "You are a creative poet. Write beautiful, evocative poems."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=0.9,
        stream=False
    )
    
    return completion.choices[0].message.content
```

```python
# Generate a poem about nature
prompt = "Write a poem about a peaceful forest at dawn"
poem = generate_poem(prompt)
print(poem)
```

```python
import OpenAI from 'openai';

const openai = new OpenAI({
  apiKey: '$NVIDIA_API_KEY',
  baseURL: 'https://integrate.api.nvidia.com/v1',
})

async function main() {
  const completion = await openai.chat.completions.create({
    model: "google/gemma-2-2b-it",
    messages: [{"role":"user","content":""}],
    temperature: 0.2,
    top_p: 0.7,
    max_tokens: 1024,
    stream: false,
  })
   
  process.stdout.write(completion.choices[0]?.message?.content);
  
}

main();
```
